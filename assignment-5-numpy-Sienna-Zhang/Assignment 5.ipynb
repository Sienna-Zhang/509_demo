# Cosin Similarity
Cosine similarity measures the similarity between two high dimensional vectors. It is widely-used in applications such as clustering tasks in machine learning, building recommendation systems for e-commerce companies. See more background on cosine similarity in [Wikipedia](https://en.wikipedia.org/wiki/Cosine_similarity). 

Write a function named "cosine_similarity". The function takes two 1D numpy arrays as inputs, and returns their cosine similarity. 

**Note**: You are allowed to use np.dot() for inner product, np.sum() for summation, np.sqrt() for calculating square root. Do not use other built-in functions from Numpy such as np.linalg.norm(). 


### The formula is:

\[
\text{cosine\_similarity}(A, B) = \frac{A \cdot B}{\|A\| \times \|B\|}
\]
### Step 1: Compute the dot product  
Use `np.dot(a, b)` to calculate \( A \cdot B \).


````python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

dot_product = np.dot(a, b)
dot_product
````

### Step 2: Compute the magnitude of each vector  
We use `np.sum()` to get the sum of squares, then `np.sqrt()` for the square root.


````python
magnitude_a = np.sqrt(np.sum(a ** 2))
magnitude_b = np.sqrt(np.sum(b ** 2))
magnitude_a, magnitude_b
````

### Step 3: Calculate cosine similarity  
Divide the dot product by the product of magnitudes.


````python
cosine_similarity_value = dot_product / (magnitude_a * magnitude_b)
cosine_similarity_value
````

### Step 4: Define the function  
Define a function `cosine_similarity(a, b)` that performs all the steps above.


````python
def cosine_similarity(a: np.ndarray, b: np.ndarray) -> float:
    """
    Calculate the cosine similarity between two 1D numpy arrays.

    Parameters :

    a : np.ndarray
        First input vector.
    b : np.ndarray
        Second input vector.

    Returns :

    float
        Cosine similarity value between a and b.
    """
    dot_product = np.dot(a, b)
    magnitude_a = np.sqrt(np.sum(a ** 2))
    magnitude_b = np.sqrt(np.sum(b ** 2))
    cosine_sim = dot_product / (magnitude_a * magnitude_b)
    return cosine_sim
````

### Step 5: Test the function
Run the function with example arrays.


````python
result = cosine_similarity(np.array([1, 2, 3]), np.array([4, 5, 6]))
print("Cosine Similarity:", result)
````

### Step 3: Define the function `count_max_streak`

1. Create a Boolean array of price increases.
2. Loop through it to count consecutive `True` values.
3. Keep track of the maximum streak length.


````python
def count_max_streak(prices: np.ndarray) -> int:
    """
    Count the maximum number of consecutive days
    when the stock price increased.

    Parameters :

    prices : np.ndarray
        1D array of daily closing stock prices.

    Returns :

    int
        The maximum streak (number of consecutive increases).
    """
    # Boolean array: True if today's price > yesterday's
    increase = prices[1:] > prices[:-1]
    
    max_streak = 0
    current_streak = 0
    
    for inc in increase:
        if inc:
            current_streak += 1
            max_streak = max(max_streak, current_streak)
        else:
            current_streak = 0
    
    return max_streak
````

### Step 1 — Generate example data

and an initial infection probability vector `p0`.


````python
import numpy as np

N = 6
np.random.seed(20)

matrix = np.random.rand(N, N)
symmetric_matrix = (matrix + matrix.T) / 2
W = symmetric_matrix / symmetric_matrix.sum(axis=1, keepdims=True)

p0 = np.random.rand(N, 1)  # initial infection probabilities
p0[:5], W.shape
````

Run several updates to see how probabilities evolve.

### Step 5 — Define the reusable function `simulate_disease`


````python
def simulate_disease(p0: np.ndarray, connection_matrix: np.ndarray, steps: int) -> np.ndarray:
    """
    Simulate disease propagation over a network using p_{t+1} = W p_t.

    Parameters :

    p0 : np.ndarray
        Initial infection probabilities, shape (N,) or (N, 1).
    connection_matrix : np.ndarray
        Symmetric, row-stochastic, non-negative matrix W of shape (N, N).
    steps : int
        Number of update steps (prediction horizon), must be >= 0.

    Returns :

    np.ndarray
        Final infection probabilities as a 1-D array of shape (N,).
    """
    # Convert inputs to proper numeric arrays
    p = np.asarray(p0, dtype=float).reshape(-1)      # flatten to (N,)
    W = np.asarray(connection_matrix, dtype=float)   # ensure float array

    # Validate dimensions
    if W.ndim != 2 or W.shape[0] != W.shape[1]:
        raise ValueError("connection_matrix must be square (N, N).")
    if W.shape[0] != p.shape[0]:
        raise ValueError("len(p0) must equal W.shape[0].")
    if np.any(W < 0):
        raise ValueError("connection_matrix must be non-negative.")
    if steps < 0:
        raise ValueError("steps must be a non-negative integer.")

    # Normalize rows to make matrix row-stochastic (sums = 1)
    row_sums = W.sum(axis=1, keepdims=True)
    if np.any(row_sums <= 0):
        raise ValueError("Each row must have a positive sum.")
    W = W / row_sums

    # Iterative propagation
    for _ in range(int(steps)):
        p = W @ p
        p = np.clip(p, 0.0, 1.0)  # ensure probabilities stay within [0, 1]

    return p
````

###  Step 1 — Define `generate_sine` function
This helper function creates a sine wave for a given frequency and duration.


````python
import numpy as np
import sounddevice as sd   

def generate_sine(t_duration, f0, fs=8000):
    """
    Generate a sine wave of given frequency and duration.

    Parameters :

    t_duration : float
        Total duration of the note in seconds.
    f0 : float
        Frequency of the sine wave (Hz).
    fs : int
        Sampling rate in samples per second.

    Returns :

    np.ndarray
        1D NumPy array of the sine wave amplitudes.
    """
    # Create time samples
    t = np.linspace(0, t_duration, int(fs * t_duration), endpoint=False)
    # Generate the sine waveform
    y = np.sin(2 * np.pi * f0 * t)
    return y
````

### Step 2 — Define the note frequencies for A major scale
Start from base note A = 220 Hz and use equal-tempered scale intervals.


````python
fs = 8000  # Sampling rate

# Define each note in the A major scale
Note_A       = generate_sine(0.5, 220, fs)
Note_B       = generate_sine(0.5, 220 * 2**(2/12), fs)
Note_Cs      = generate_sine(0.5, 220 * 2**(4/12), fs)
Note_D       = generate_sine(0.5, 220 * 2**(5/12), fs)
Note_E       = generate_sine(0.5, 220 * 2**(7/12), fs)
Note_Fs      = generate_sine(0.5, 220 * 2**(9/12), fs)
Note_Gs      = generate_sine(0.5, 220 * 2**(11/12), fs)
Note_A_high  = generate_sine(0.5, 440, fs)
````

### Step 3 — Prepare a music sheet
Each tuple represents (note_name, duration).  
We’ll later map each name to its waveform.


````python
music_sheet = [
    ("Note_Cs", 0.5),
    ("Note_D", 0.5),
    ("Note_B", 0.5),
    ("Note_A_high", 1.5),
    ("Note_A_high", 0.5),
    ("Note_Gs", 0.5),
    ("Note_Fs", 0.5),
    ("Note_Gs", 0.5),
    ("Note_E", 0.75)
]
````

### Step 4 — Define the `compose_music` function
This function takes a music sheet and generates the full melody by concatenating notes.


````python
def compose_music(music_sheet, fs=8000):
    """
    Compose a piece of music from a list of (note_name, duration) pairs.

    Parameters :

    music_sheet : list of tuples
        Each tuple contains (note_name: str, duration: float)
    fs : int
        Sampling rate for audio generation.

    Returns :

    np.ndarray
        1D NumPy array representing the full music waveform.
    """
    # Define the frequency mapping for the A major scale
    base_freq = 220
    note_freqs = {
        "Note_A": base_freq,
        "Note_B": base_freq * 2**(2/12),
        "Note_Cs": base_freq * 2**(4/12),
        "Note_D": base_freq * 2**(5/12),
        "Note_E": base_freq * 2**(7/12),
        "Note_Fs": base_freq * 2**(9/12),
        "Note_Gs": base_freq * 2**(11/12),
        "Note_A_high": 440
    }

    # Initialize an empty list to store generated notes
    music_waveform = []

    # Loop through the music sheet
    for note_name, duration in music_sheet:
        if note_name not in note_freqs:
            raise ValueError(f"Unknown note: {note_name}")
        frequency = note_freqs[note_name]

        # Generate sine wave for this note
        note_wave = generate_sine(duration, frequency, fs)

        # Append to the full music waveform
        music_waveform.append(note_wave)

    # Concatenate all note waveforms into one long signal
    music_waveform = np.concatenate(music_waveform)

    return music_waveform
````

### Step 5 — Generate and play the music
Use `compose_music()` to create the waveform and `sounddevice.play()` to listen.


````python
# Generate the waveform
melody = compose_music(music_sheet, fs)

# Play the sound (comment out if sounddevice is unavailable)
sd.play(melody, fs)

# Wait until playback finishes
sd.wait()
````

# Linear Regression
- You are given a straight line learned from linear regression, denoted as $\hat{y}=ax+b$. Your task is to use a Python function named `eval_predict` to assess prediction quality over a dataset. Please first define the metric being used for assessment. Then let your program calculate and return the defined metric value.
- To learn a high-quality straight line in the form of $\hat{y}=ax+b$, we need to find optimal values for $a$
 and $b$. In what follows, you will perform grid search. You’ll try many pairs on a rectangular grid and pick the one with the best metric (defined in Problem 7) on the dataset. For example, consider a ∈ {0.0, 0.5, 1.0} and b ∈ {-1.0, 0.0, 1.0}, you evaluate all 9 combinations, compute the metric for each, and pick the best. Write a Python function named `grid_search` to find best pairs for given options of $a$ and $b$.

### Overview:
1. **Evaluate** how good a given line (with parameters `a` and `b`) is on a dataset.  
   → Implement `eval_predict(a, b, x, y)`
2. **Find the best parameters** `a` and `b` by testing all combinations from given lists.  
   → Implement `grid_search(a_list, b_list, x, y)`

### Step 1 — Define the evaluation metric
Use **Mean Squared Error (MSE)** as our metric:

\[
MSE = \frac{1}{n} \sum (y_i - \hat{y}_i)^2
\]

The smaller the MSE, the better the model.


````python
import numpy as np

def eval_predict(a, b, x, y):
    """
    Evaluate prediction quality using Mean Squared Error (MSE).

    Parameters:

    a : float
        Slope of the line.
    b : float
        Intercept of the line.
    x : np.ndarray
        Input feature values.
    y : np.ndarray
        True target values.

    Returns:

    float
        Mean Squared Error between predicted and true values.
    """
    # Compute predictions
    y_hat = a * x + b

    # Compute MSE
    mse = np.mean((y - y_hat) ** 2)
    return mse
````

### Step 2 — Grid Search Concept
Perform a **grid search**:
- Loop through every combination of `a` and `b`
- Evaluate each using `eval_predict()`
- Keep track of the combination with the smallest MSE


````python
def grid_search(a_list, b_list, x, y):
    """
    Perform grid search to find the best (a, b) pair that minimizes MSE.

    Parameters:

    a_list : list
        Candidate values for slope 'a'.
    b_list : list
        Candidate values for intercept 'b'.
    x : np.ndarray
        Input feature values.
    y : np.ndarray
        True target values.

    Returns:

    tuple
        (best_a, best_b, best_mse)
    """
    best_a, best_b = None, None
    best_mse = float('inf')  # initialize with infinity

    # Try every combination of (a, b)
    for a in a_list:
        for b in b_list:
            mse = eval_predict(a, b, x, y)
            if mse < best_mse:
                best_mse = mse
                best_a = a
                best_b = b

    return best_a, best_b, best_mse
````

### Step 3 — Test the functions
We’ll create a small synthetic dataset following approximately:
\[
y = 2x + 1
\]
and run our grid search to find the best parameters.


````python
# Example dataset
x = np.array([0, 1, 2, 3, 4])
y = np.array([1, 3, 5, 7, 9])  # True model: y = 2x + 1

# Candidate parameter values
a_values = np.arange(0.0, 3.0, 0.5)
b_values = np.arange(-1.0, 2.0, 0.5)

# Run grid search
best_a, best_b, best_mse = grid_search(a_values, b_values, x, y)

print("Best a:", best_a)
print("Best b:", best_b)
print("Best MSE:", best_mse)
````

### Step 1 — Username validation (`validate_username`)
Rules:
- Return `False` if empty, length < 3 or > 20.
- Return `False` if it contains anything besides letters, digits, underscore `_`, dot `.`, or hyphen `-`.
- Return `True` otherwise.


````python
import re

def validate_username(username: str) -> bool:
    """
    Validate a username:
      - length in [3, 20]
      - only letters, digits, underscore, dot, hyphen
    """
    if username is None:
        return False
    username = username.strip()
    if not (3 <= len(username) <= 20):
        return False
    pattern = r'^[A-Za-z0-9_.-]+$'
    return re.fullmatch(pattern, username) is not None
````

- The domain must contain at least one "." and not start/end with "."
- Return `True` if valid, else `False`


````python
def validate_email(email: str) -> bool:
    """
    Simple email validation per prompt:
      - contains '@'
      - use partition('@') to split into local and domain
      - domain contains at least one '.'
      - domain does not start or end with '.'
    """
    if email is None:
        return False
    local, sep, domain = email.partition("@")
    if sep == "":  # no '@'
        return False
    if not local or not domain:
        return False
    if "." not in domain:
        return False
    if domain.startswith(".") or domain.endswith("."):
        return False
    return True
````

### Step 3 — Phone normalization (simplified) (`normalize_phone`)
Rules:
- Remove all non-digit characters.
- If the original input does **not** start with "+", prepend `default_cc` (e.g., "+1").
- Check digit count (excluding '+') is between 9 and 15.
- Return normalized string like "+14155551212" or "Invalid".


````python
def normalize_phone(phone: str, default_cc: str = "+1") -> str:
    """
    Normalize a phone number to a simplified E.164-like string.

    Steps:
      1) Detect whether the *original* string starts with '+'
      2) Remove all non-digit characters to get digits only
      3) If original did not start with '+', prepend default_cc
      4) Validate total digit count in [9, 15] (excluding '+')
      5) Return normalized '+<digits>' or 'Invalid'
    """
    if phone is None:
        return "Invalid"
    s = phone.strip()
    has_plus_prefix = s.startswith("+")
    digits = "".join(ch for ch in s if ch.isdigit())

    if not digits:
        return "Invalid"

    if has_plus_prefix:
        normalized = "+" + digits
    else:
        # Ensure default_cc looks like '+<ccdigits>'
        if not (default_cc and default_cc.startswith("+") and default_cc[1:].isdigit()):
            return "Invalid"
        normalized = default_cc + digits

    # Validate digit count (exclude '+')
    digit_count = len(normalized) - 1
    if 9 <= digit_count <= 15:
        return normalized
    return "Invalid"
````

### Step 4 — Sign-up aggregator (`validate_signup`)
- Call all validators
- Collect errors
- Produce a `normalized` object:
  - username: original if valid (or keep original even if invalid for visibility)
  - email: lowercased if valid
  - phone: normalized string or "Invalid"
  - country: uppercased if present
- Return:
```json
{
  "ok": true/false,
  "errors": {...},
  "normalized": {...}
}


````python

def validate_signup(user_info: dict) -> dict:
    """
    Validate and normalize sign-up info.

    Expected keys in user_info:
      - username (str)
      - email (str)
      - password (str)  # not validated beyond presence in this prompt
      - phone (str)
      - country (str)
    """
    errors = {}
    normalized = {}

    # Username
    username = user_info.get("username", "")
    if not validate_username(username):
        errors["username"] = "Invalid username (length or characters)."
    normalized["username"] = username

    # Email
    email = user_info.get("email", "")
    if not validate_email(email):
        errors["email"] = "Invalid email format."
        normalized["email"] = email
    else:
        normalized["email"] = email.lower()

    # Phone
    phone = user_info.get("phone", "")
    # Use default_cc based on country if desired; here we stick with default "+1"
    normalized_phone = normalize_phone(phone, default_cc="+1")
    if normalized_phone == "Invalid":
        errors["phone"] = "Invalid phone number."
    normalized["phone"] = normalized_phone

    # Country
    country = user_info.get("country", "")
    normalized["country"] = country.upper() if isinstance(country, str) else country

    # Password (prompt does not specify rules; ensure presence only if you want)
    pwd = user_info.get("password", "")
    if pwd is None or pwd == "":
        errors["password"] = "Password is required."

    ok = (len(errors) == 0)
    return {"ok": ok, "errors": errors, "normalized": normalized}
````
