# Password Meter

- Implement `has_run_of_k(s: str, k: int) -> bool`. Return True if any char repeats ≥ k times consecutively.

- Implement `password_score(pw: str) -> tuple[str, int]`. Apply your previous Password Strength rules; return `(label, score)`.

````python
def has_run_of_k(s: str, k: int) -> bool:
    """Return True if any character repeats ≥ k times consecutively."""
    if k <= 1:
        return True
    for i in range(len(s) - k + 1):
        if all(s[i] == s[i + j] for j in range(k)):
            return True
    return False


def password_score(pw: str) -> tuple[str, int]:
    """Apply previous Password Strength rules; return (label, score)."""
    specials = "!@#$%^&*"
    score = 0

    # Length rule
    if len(pw) >= 12:
        score += 2
    elif len(pw) >= 8:
        score += 1

    # Character types
    has_lower = any(ch.islower() for ch in pw)
    has_upper = any(ch.isupper() for ch in pw)
    has_digit = any(ch.isdigit() for ch in pw)
    has_special = any(ch in specials for ch in pw)

    if has_lower and has_upper:
        score += 1
    if has_digit:
        score += 1
    if has_special:
        score += 1

    # Space penalty
    if any(ch.isspace() for ch in pw):
        score -= 2

    # Repeated run penalty
    if has_run_of_k(pw, 3):
        score -= 3

    # Determine label
    if score < 3:
        strength = "weak"
    elif score <= 4:
        strength = "okay"
    else:
        strength = "strong"

    return strength, score


# Example usage
passwords = ["P@ssw0rd", "hunter2", "CorrectHorseBatteryStaple", "abcde Fgh", "A!!!A!!!A!!!"]

for pwd in passwords:
    strength, score = password_score(pwd)
    print(f"Checking: {pwd}")
    print(f"Score: {score} | Strength: {strength}\n")
````

# Tic-Tac-Toe Winner

Implement `find_winner(board: list[list[str]]) -> str`. Board uses "X", "O", or "". Return "X", "O", "Draw", or "Pending". Here "Pending" indicates that game is not finished yet.


````python
def find_winner(board: list[list[str]]) -> str:
    """
    Determine the winner of a Tic-Tac-Toe board.
    Board is a 3x3 list using "X", "O", or "" (empty string).
    Return "X", "O", "Draw", or "Pending".
    """

    n = len(board)

    # Check rows
    for i in range(n):
        if board[i][0] != "" and board[i][0] == board[i][1] == board[i][2]:
            return board[i][0]

    # Check columns
    for j in range(n):
        if board[0][j] != "" and board[0][j] == board[1][j] == board[2][j]:
            return board[0][j]

    # Check diagonals
    if board[0][0] != "" and board[0][0] == board[1][1] == board[2][2]:
        return board[0][0]
    if board[0][2] != "" and board[0][2] == board[1][1] == board[2][0]:
        return board[0][2]

    # Check for pending (empty cell)
    for row in board:
        if "" in row:
            return "Pending"

    # If no empty cell and no winner 
    return "Draw"


# Example
board1 = [['O', 'X', 'X'],
          ['X', 'X', 'O'],
          ['O', 'X', 'X']]

board2 = [['O', 'O', 'O'],
          ['X', 'X', ''],
          ['', '', '']]

board3 = [['O', 'X', 'O'],
          ['X', 'O', 'X'],
          ['X', 'O', 'X']]

board4 = [['O', 'X', ''],
          ['X', '', 'O'],
          ['', '', '']]

print("Board 1:", find_winner(board1)) 
print("Board 2:", find_winner(board2))
print("Board 3:", find_winner(board3))
print("Board 4:", find_winner(board4))
````

# Tic-Tac-Toe Board
- Write a function named `draw_board` to print out a game board to the user. The function consumes a game board as input, and prints out a square board of appropriate size based on the board given. For this problem, you can assume the board is represented as a list of lists (see Assignment of Week 3 for an example). An example of 2x2 board is as follows:
    ```cmd
    --- --- 
    | X | O | 
    --- --- 
    |   | X |
    --- --- 
    ```

- Write a function named `update_board`. The function should allow players to take turns marking the board, and update the board as the game progresses. Assume player ID 1 will place 'X' and player ID 2 will place 'O' on board. The function should:
  - If the board is non-empty, then figure out who should mark in this turn based on the current board. If there are an equal number of 'X' and 'O' marks, it's player 1's turn, otherwise it's player 2's
  - Check if there is empty space on the board for a player to add 'X' or 'O'. If there is an empty space on board, prompt the user to input the row index and column index for the mark to be placed on board. 
  - Check if the input indices are valid or not (whether the indices refer to an empty space). If so, update the board and return the updated version. If not, re-prompt the user for another pair row and column indices.

````python
def is_empty_cell(val: str) -> bool:
    """Return True if cell is empty ('' or whitespace)."""
    return val is None or (isinstance(val, str) and val.strip() == "")


def draw_board(board: list[list[str]]) -> None:
    """Print the game board."""
    n = len(board)
    for i in range(n):
        print("--- " * n)
        for j in range(n):
            cell = " " if is_empty_cell(board[i][j]) else board[i][j]
            print(f"| {cell} ", end="")
        print("|")
    print("--- " * n)


def update_board(board: list[list[str]]) -> list[list[str]]:
    """Update the board for the next move."""
    n = len(board)

    # Normalize board: convert empty cells to ""
    for i in range(n):
        for j in range(n):
            if is_empty_cell(board[i][j]):
                board[i][j] = ""

    # Count marks to decide turn
    count_x = sum(row.count('X') for row in board)
    count_o = sum(row.count('O') for row in board)
    player = 1 if count_x == count_o else 2
    mark = 'X' if player == 1 else 'O'

    # Check if any empty cell remains
    if not any(is_empty_cell(cell) for row in board for cell in row):
        print("No empty spaces left. Game over.")
        return board

    print(f"Player {player}'s turn ({mark})")

    # Input until valid
    while True:
        try:
            row = int(input(f"Enter row index (0-{n-1}): "))
            col = int(input(f"Enter column index (0-{n-1}): "))

            if row < 0 or row >= n or col < 0 or col >= n:
                print("Invalid index.")
                continue

            if not is_empty_cell(board[row][col]):
                print("Cell taken.")
                continue

            board[row][col] = mark
            break

        except ValueError:
            print("Enter valid integers.")

    return board


if __name__ == "__main__":
    board = [[" ", " ", " "],
             ["X", "O", "X"],
             ["X", "O", "X"]]

    while True:
        draw_board(board)
        board = update_board(board)
        draw_board(board)
        cont = input("Continue? (y/n): ").lower()
        if cont != "y":
            break
````

# Path Planning for Vacuum Robot
Consider a vacuum robot moving in a grid world as seen in previous assignments. Our goal is to find a path for the robot to move from starting point, denoted as `start`, to destination, denoted as `dest`. During the movements, the robot is required to avoid walls `#`.

The grid world is an $n\times n$ grid. You can move up, right, down, left. No diagonal movement is allowed.
A path is a sequence of adjacent cells from `start` to `dest` that never goes through `#`.

Your task is to **apply recursion** to find a valid path for the robot to move from `start` to `dest`. Your program should return the path starting with `start` and ending with `dest`. For this problem, you do NOT need to consider the optimality of path found, i.e., the shortest path.
Please structure your program using functions, and write a main program to call the functions.

Hint to initiate your thinking:
One algorithm to find the path is called depth-first search (DFS) with backtracking:
- Try a direction.
- If robot gets stuck, backtrack (undo the step) and try the next direction.
- Stop when you reach `dest`.


````python
# Directions: up, right, down, left
DIRS = [(-1, 0), (0, 1), (1, 0), (0, -1)]


def find_start_and_dest(grid: list[str]) -> tuple[tuple[int, int], tuple[int, int]]:
    """Find coordinates of 'S' (start) and 'D' (destination)."""
    start = dest = None
    for i in range(len(grid)):
        for j in range(len(grid[0])):
            if grid[i][j] == 'S':
                start = (i, j)
            elif grid[i][j] == 'D':
                dest = (i, j)
    return start, dest


def is_valid(grid: list[str], r: int, c: int, visited: set) -> bool:
    """Check if cell (r, c) is within bounds, not a wall, and not visited."""
    n = len(grid)
    return (
        0 <= r < n and 0 <= c < len(grid[0])
        and grid[r][c] != '#'
        and (r, c) not in visited
    )


def dfs(grid: list[str], r: int, c: int, dest: tuple[int, int], path: list, visited: set) -> bool:
    """Recursive DFS search from (r, c) to destination."""
    if (r, c) == dest:           # reached destination
        path.append((r, c))
        return True

    visited.add((r, c))
    path.append((r, c))

    for dr, dc in DIRS:
        nr, nc = r + dr, c + dc
        if is_valid(grid, nr, nc, visited):
            if dfs(grid, nr, nc, dest, path, visited):
                return True  # found path

    # Backtrack: remove last step if dead end
    path.pop()
    return False


def find_path(grid: list[str]) -> list[tuple[int, int]]:
    """Find a valid path from S to D avoiding walls."""
    start, dest = find_start_and_dest(grid)
    if not start or not dest:
        print("Missing S or D on grid.")
        return []

    path, visited = [], set()
    dfs(grid, start[0], start[1], dest, path, visited)
    return path


# Example grids

if __name__ == "__main__":
    grid1 = [
        "#####",
        "#S..#",
        "#.#.#",
        "#..D#",
        "#####",
    ]

    path = find_path(grid1)
    if path:
        print("Path found:")
        print(path)
    else:
        print("No valid path found.")
````
