# FizzBuzz
Write a program that prints the numbers from 1 to 100. But for multiples of three print “Fizz” instead of the number and for the multiples of five print “Buzz”. For numbers that are multiples of both three and five print “FizzBuzz”.

````python
for i in range(1,101):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 5 == 0:
        print("Buzz")
    elif i % 3 ==0:
        print("Fizz")
    else:
        print(i)
````

# Tic-Tac-Toe Game Winner
Given a board of a tic-tac-toe game, find the winner. A sample board is given as follows:
```py
board = [['O', 'X', 'X'],
         ['X', 'X', 'O'],
         ['O', 'X', 'X']]
```


````python
board = [['O', 'X', 'X'],
         ['X', 'X', 'O'],
         ['O', 'X', 'X']]

winner = None 
n = 3   

for i in range(n):
    if board[i][0] == board[i][1] == board[i][2]:
        winner = board[i][0]


for j in range(n):
    if board[0][j] == board[1][j] == board[2][j]:
        winner = board[0][j]


if board[0][0] == board[1][1] == board[2][2]:
    winner = board[0][0]
if board[0][2] == board[1][1] == board[2][0]:
    winner = board[0][2]


if winner == 'X' or winner == 'O':
    print("Winner:", winner)
else:
    print("No winner")
````

# Password Strenghth Meter
Write a Python program that reads passwords from users. For the given password, tell users whether the password is **weak**, **okay**, or **strong**. Here are the scoring rules for a password:

- Length
  - +2 if length ≥ 12
  - +1 if length is 8–11
  - +0 otherwise
- Character variety
  - +1 if it has both lowercase [a-z] and uppercase [A-Z]
  - +1 if it has at least one digit [0-9]
  - +1 if it has at least one special from this set only: !@#$%^&*
- Penalties
  - −2 if it contains any whitespace (space, tab, etc.)
  - −3 if it contains three or more identical characters in a row (e.g., aaa, 1111, ***)
- Final Result
  - weak if score < 3
  - okay if 3 ≤ score ≤ 4
  - strong if score ≥ 5

Test your program with the following passwords:
- P@ssw0rd
- hunter2
- CorrectHorseBatteryStaple
- abcde Fgh
- A!!!A!!!A!!!

**Hint**: You may find `str.islower()`, `str.isupper()`, `str.isdigit()`, and `str.isspace()` useful.

````python
passwords = ["P@ssw0rd", "hunter2", "CorrectHorseBatteryStaple", "abcde Fgh", "A!!!A!!!A!!!"]

specials = "!@#$%^&*"

for pwd in passwords:
    score = 0
    print("Checking:", pwd)

    if len(pwd) >= 12:
        score += 2
    elif len(pwd) >= 8:
        score += 1

    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False

    for ch in pwd:
        if ch.islower():
            has_lower = True
        elif ch.isupper():
            has_upper = True
        elif ch.isdigit():
            has_digit = True
        elif ch in specials:
            has_special = True

    if has_lower and has_upper:
        score += 1
    if has_digit:
        score += 1
    if has_special:
        score += 1

    has_space = False
    for ch in pwd:
        if ch.isspace():
            has_space = True
    if has_space:
        score -= 2

    for i in range(len(pwd) - 2):
        if pwd[i] == pwd[i+1] == pwd[i+2]:
            score -= 3
            break   

    
    if score < 3:
        strength = "weak"
    elif score <= 4:
        strength = "okay"
    else:
        strength = "strong"

    print("Score:", score, "| Strength:", strength)
    print()
````

# Vacuum Robot

You get an $n\times n$ map made of:
- \# = wall (can’t go through)
- . = empty floor (can move)
- S = where the robot starts (treated like empty floor)
You’re told which way the robot is facing (N, E, S, or W).

You get a string of commands:
- L turn left 90°
- R turn right 90°
- F try to move one step forward
If a forward step would hit a wall or leave the map, the robot remains where it is. Its health condition decreases by 1.
After all commands, print: row col orientation (rows/cols are 1-based, top-left is (1,1)), as well as robot's health condition.
**Note**: The minimum health condition is 0. Once the robot's health reaches 0, it cannot accept commands any more. 

Test your program with the following grids and commands:

- Grid 1
    ```py
    #####
    #...#
    #.#S#
    #...#
    #####
    ```
    Commands: FFLFFRFFF; assume the robot initially facing east (E) with initial health 3.

- Grid 2
    ```py
    #######
    #..#..#
    #.#.#S#
    #.#...#
    #..####
    #.....#
    #######
    ```
    Commands: FLLFFFRFFLFFRFF; assume the robot initially facing west (W) with initial health 10.

````python
dirs = ['N', 'E', 'S', 'W']  
drdc = {'N': (-1, 0), 'E': (0, 1), 'S': (1, 0), 'W': (0, -1)}

# Grid 1
grid = [
    "#####",
    "#...#",
    "#.#S#",
    "#...#",
    "#####",
]
commands = "FFLFFRFFF"
facing = 'E'    
health = 3  


r = c = None
for i in range(len(grid)):
    for j in range(len(grid[0])):
        if grid[i][j] == 'S':
            r, c = i, j


for cmd in commands:
    if health <= 0:
        break
    if cmd == 'L':
        facing = dirs[(dirs.index(facing) - 1) % 4]
    elif cmd == 'R':
        facing = dirs[(dirs.index(facing) + 1) % 4]
    elif cmd == 'F':
        dr, dc = drdc[facing]
        nr, nc = r + dr, c + dc
    
        if 0 <= nr < len(grid) and 0 <= nc < len(grid[0]) and grid[nr][nc] != '#':
            r, c = nr, nc
        else:
            health = max(0, health - 1) 
  


print(r + 1, c + 1, facing, health)  

# Grid 2
grid = [
    "#######",
    "#..#..#",
    "#.#.#S#",
    "#.#...#",
    "#..####",
    "#.....#",
    "#######",
]
commands = "FLLFFFRFFLFFRFF"
facing = 'W'
health = 10


r = c = None
for i in range(len(grid)):
    for j in range(len(grid[0])):
        if grid[i][j] == 'S':
            r, c = i, j


for cmd in commands:
    if health <= 0:
        break
    if cmd == 'L':
        facing = dirs[(dirs.index(facing) - 1) % 4]
    elif cmd == 'R':
        facing = dirs[(dirs.index(facing) + 1) % 4]
    elif cmd == 'F':
        dr, dc = drdc[facing]
        nr, nc = r + dr, c + dc
        if 0 <= nr < len(grid) and 0 <= nc < len(grid[0]) and grid[nr][nc] != '#':
            r, c = nr, nc
        else:
            health = max(0, health - 1)
    

print(r + 1, c + 1, facing, health) 
````
